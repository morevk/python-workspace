```python
'''
Basics & Creation
Empty List: my_list = []

Initialization: items = ["apple", "banana", 10, 3.14]

List Length: len(items) returns the number of elements.
'''

# Create and print a list
fruits = ["apple", "banana", "cherry"]
print(f"List: {fruits}")
print(f"Length: {len(fruits)}")
```

```python
'''
Accessing & Slicing
Lists use zero-based indexing.

Access: items[0] (first) or items[-1] (last).

Slice: items[start:stop:step]

items[1:3] (index 1 to 2)

items[::-1] (reverses the list)
'''
numbers = [0, 10, 20, 30, 40, 50]

print(f"First element: {numbers[0]}")
print(f"Last element: {numbers[-1]}")
print(f"Slice (1 to 4): {numbers[1:4]}")
print(f"Reversed: {numbers[::-1]}")
```

```python
'''
 Modifying ListsLists are mutable, meaning you can change them in place.
 OperationMethod / Syntax
 Add to end   .append(item)
 Insert       .insert(index, item)
 Extend       .extend(iterable)
 Remove by value  .remove(item)
 Remove by index  .pop(index)
 Clear             .clear()
'''
colors = ["red", "blue"]

# Adding
colors.append("green")
colors.insert(1, "yellow")
print(f"After adding: {colors}")

# Removing
popped_item = colors.pop() # Removes last item
colors.remove("yellow")
print(f"Popped: {popped_item}")
print(f"After removing: {colors}")
```

```python
'''
Searching & Sorting
in operator: Check if item exists.

.sort(): Sorts the list in place.

sorted(list): Returns a new sorted list.
'''

tech = ["Python", "C++", "Java", "Assembly"]

# Searching
print(f"Is 'Python' present? {'Python' in tech}")

# Sorting
tech.sort()
print(f"Sorted list: {tech}")

# Count & Index
print(f"Index of 'Java': {tech.index('Java')}")
```

```python
'''
List Comprehensions
A concise way to create lists.
Syntax: [expression for item in iterable if condition]
'''
# Create a list of squares for even numbers between 0 and 9
squares = [x**2 for x in range(10) if x % 2 == 0]
print(f"Even squares: {squares}")
```

```python
'''
Heterogeneous Data
Lists can store multiple data types simultaneously because they store references to objects.
'''
mixed_data = [101, "Firmware", 5.5, True, [1, 2]]

for item in mixed_data:
    print(f"Value: {item} \t Type: {type(item)}")
```
