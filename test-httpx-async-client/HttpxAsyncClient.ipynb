```python
import asyncio
import httpx

async def fetch_size(client: httpx.AsyncClient, url: str) -> None:
    try:
        # Perform the GET request
        response = await client.get(url, follow_redirects=True)
        # Calculate size in bytes (length of the raw bytes content)
        size = len(response.content)
        print(f"Success: {url} -> {size} bytes (Status: {response.status_code})")
    except httpx.HTTPError as e:
        print(f"Failed : {url} -> Error: {e}")

async def main():
    urls = [
        "https://www.python.org",
        "https://www.wikipedia.org",
        "https://httpbin.org/delay/1"  # Good for testing concurrency
    ]
    
    # Use AsyncClient as a context manager to handle connection pooling
    async with httpx.AsyncClient() as client:
        # Create a list of coroutines
        tasks = [fetch_size(client, url) for url in urls]
        
        # Fire them off concurrently and wait for all to complete
        await asyncio.gather(*tasks)

if __name__ == "__main__":
    # Run the main async loop
    # [vishal]: Error in jupyter notebooks
    # asyncio.run(main()) 
    await main()
```
