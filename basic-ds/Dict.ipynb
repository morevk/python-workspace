```python
# Create and print a dictionary
device = {"host": "192.168.1.10", "port": 8080, "secure": True}
print(f"Dictionary: {device}")
print(f"Total parameters: {len(device)}")
```

```python
config = {"timeout": 30, "retries": 3}

# Direct access
print(f"Timeout duration: {config['timeout']}")

# Safe access with .get()
print(f"Buffer size (default fallback): {config.get('buffer_size', 4096)}")
# print(config['buffer_size']) # This would raise a KeyError!
```

```python
status = {"status_code": 200, "message": "OK"}

# Adding and updating
status["version"] = "v2.0"
status["message"] = "Success" 
print(f"After updates: {status}")

# Removing
msg = status.pop("message")
print(f"Popped message: {msg}")
print(f"Final state: {status}")
```

```python
metrics = {"cpu_usage": 45.2, "mem_usage": 61.8, "temp": 42.0}

# Iterating over key-value pairs
for metric, value in metrics.items():
    print(f"Telemetry -> {metric}: {value}%")
```

```python
# Create a dictionary mapping numbers to their cubes, keeping only even cubes
cubes = {x: x**3 for x in range(1, 6) if (x**3) % 2 == 0}
print(f"Even cubes dictionary: {cubes}")
```

```python
base_cfg = {"debug": False, "log_level": "INFO"}
local_cfg = {"debug": True, "port": 9000}

# Merge into a completely new dictionary (Right side overwrites left side keys)
final_cfg = base_cfg | local_cfg
print(f"Merged config: {final_cfg}")
```

```python
for i in range(1,6):
    print(f"i => {i}")
```

```python
base_config = {"host": "localhost", "port": 8080}
auth_config = {"user": "admin", "password": "secret_password"}

# Merge them together
full_config = {**base_config, **auth_config}

print(full_config)
# Output: {'host': 'localhost', 'port': 8080, 'user': 'admin', 'password': 'secret_password'}
```

```python
default_settings = {"theme": "light", "notifications": True}
user_settings = {"theme": "dark"}

# user_settings comes last, so its "theme" overwrites default_settings
final_settings = {**default_settings, **user_settings}

print(final_settings)
# Output: {'theme': 'dark', 'notifications': True}
```

```python
def create_profile(name, role, status):
    print(f"User: {name} | Role: {role} | Status: {status}")

user_data = {"name": "Alice", "role": "DevOps", "status": "Active"}

# Instead of: create_profile(user_data["name"], user_data["role"], user_data["status"])
create_profile(**user_data)

# Output: User: Alice | Role: DevOps | Status: Active
```

```python
def print_report(**kwargs):
    # kwargs is a standard Python dictionary containing all passed arguments
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_report(cpu=45, memory=70, disk="20% Buffered")
```
