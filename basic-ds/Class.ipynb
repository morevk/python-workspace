```python
class Agent:
    def __init__(self, name: str, tools: list[str]):
        self.name = name
        self.tools = tools
        self.history: list[dict] = []
    def remember(self, message: dict) -> None:
        self.history.append(message)
    def __repr__(self) -> str:
        return f"Agent(name={self.name!r}, tools={self.tools})"
    def show_history(self) -> None:
        print(self.history)

researcher = Agent(name="Researcher", tools=["search", "summarise"])
researcher.remember({"role": "user", "content": "Find papers on RAG"})
print(researcher)
researcher.show_history()
```
