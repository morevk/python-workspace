```python
'''
Basic Operations
Concatenation: Join strings using +.

Repetition: Repeat strings using *.

Length: Get the number of characters using len().
'''
s = "python"
print(s + " is fun")
print(s * 3)
print("Length = " + str(len(s)) )
f"Length = {len(s)}"
```

```python
'''
Indexing & Slicing
Python uses zero-based indexing. You can access specific parts of a string using [start:stop:step].
'''
print(s[0])
print(s[-1])
print(s[1:4])
print(s[:3])
print(s[::2])
print(s[::1])
print(s[::-1])
```
