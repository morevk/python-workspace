```python
from dataclasses import dataclass, field

@dataclass
class Message:
    role: str
    content: str
    tokens: int = 10
    # Use default_factory to safely handle the default list
    zip: list = field(default_factory=lambda: [17, 57])

m = Message(role="user", content="Hi")
print(f"Message -> {m}")

m = Message(role="user", content="Hi", tokens=20, zip=[10,20])
print(f"Message -> {m}")
```
